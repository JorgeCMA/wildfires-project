Cada etapa escribe su CSV en `data/processed/` y alimenta a la siguiente; `data/raw/` nunca se modifica.

Full pipeline in two blocks: **1) enrichment** (FIRMS merge → Open-Meteo weather → CLCPlus cover → CCAA and budgets) and **2) cleaning and feature engineering** (cross-sensor dedup, descriptives, imputation, temporality, holidays and final indices).
Each stage writes its CSV under `data/processed/` and feeds the next; `data/raw/` is never modified.


## 1.1 — Recopilación FIRMS y confianza unificada

Carga los CSV crudos de los tres sensores (2023 y 2024), los concatena y unifica la confianza; escribe `merged/firms_spain_merged.csv`.

```python
# Celda 1 — Recopilación de todos los datos FIRMS y corrección de la confianza
#
# `merge_viirs_modis` hace dos cosas:
#   1. Carga los CSVs crudos de los tres sensores (modis, viirs_snpp, viirs_noaa20)
#      para los años indicados (2023 y 2024) y los concatena en un único DataFrame.
#   2. Llama a `add_unified_confidence` para corregir/unificar la confianza:
#      - MODIS trae confianza NUMÉRICA (0-100) -> se convierte a categórica (l/n/h)
#        usando los rangos de configs/confidence_thresholds.yaml.
#      - VIIRS trae confianza CATEGÓRICA (l/n/h) -> se convierte a numérica (15/50/85).
#      - Los valores originales se conservan en confidence_og_num / confidence_og_cat,
#        y el resultado unificado queda en confidence_cat / confidence_num.
#
# `save_merged` escribe el resultado en data/processed/merged/. Los datos crudos
# de data/raw/ nunca se modifican (principio de inmutabilidad del pipeline).


# 1) Agrupar todos los CSVs FIRMS (VIIRS + MODIS, 2023 + 2024) y corregir la confianza
df_merged = merge_viirs_modis(country="Spain", years=[2023, 2024])

# 2) Guardar el dataset unificado en un CSV de salida (se ejecuta dentro de esta celda)
ruta_salida = save_merged(df_merged, country="Spain")

print(f"Filas unificadas: {len(df_merged):,}")
print(f"Columnas: {df_merged.shape[1]}")
print(f"Distribución de confianza unificada (confidence_cat):")
print(df_merged["confidence_cat"].value_counts().to_string())
print(f"CSV guardado en: {ruta_salida}")

df_merged.head()
```

```python
# Celda 2 — Descarga de clima histórico (Open-Meteo) por lotes / Weather download (batched)
#
# [ESP] Esta celda está COMENTADA a propósito: son ~95 lotes de 500 filas
#   en total (~2 h si se hace entero; al reanudar solo se bajan los pendientes,
#   a razón de batch_interval_seconds: 450) y gasta cuota real de la API
#   Open-Meteo. Se ejecuta fuera del notebook, sin supervisión:
#
#       python scripts/enrich_weather_batch.py
#
#   - Reanuda solo desde data/processed/enriched/firms_spain_weather_partial.csv
#     (primera fila sin temperature_2m); la respuesta HTTP cruda queda en
#     data/processed/openmeteo_response.json.
#   - Ante un 429 espera al siguiente límite minuto/hora UTC y reintenta el
#     mismo lote; el límite diario para limpio (exit 3) y al día siguiente se
#     reanuda donde quedó.
#   - Al completarse publica data/processed/enriched/firms_spain_weather.csv
#     (finalize_weather_csv), que es la entrada de la celda 3.
#   - Códigos de salida: 0 completado/max-batches, 3 límite diario,
#     4 esperas consecutivas excesivas, 1 error.
#
# [EN] This cell is intentionally COMMENTED OUT: ~95 batches of 500 rows
#   total (~2 h end-to-end; a resume only fetches the pending batches at
#   450 s interval) that spend real Open-Meteo API quota. Run it
#   outside the notebook, unattended:
#
#       python scripts/enrich_weather_batch.py
#
#   It resumes automatically from firms_spain_weather_partial.csv and, when
#   complete, publishes firms_spain_weather.csv — the input of cell 3.
#   429s are waited out; the daily limit stops cleanly (exit 3).
#
# from wildfire.enrichment.weather_batch import run_weather_enrichment
#
# status = run_weather_enrichment(
#     max_batches=None,        # None = hasta completar / until complete
#     interval_seconds=450,    # openmeteo.batch_interval_seconds
# )
# print(status)
```

## 1.3 — Dataset de clima listo para CLC

`load_weather_for_clc` usa el CSV terminado si existe; si no, el parcial filtrado a filas con `temperature_2m`.

```python
# Celda 3 — Dataset con clima listo para CLC / Weather dataset ready for CLC
#
# [ESP] load_weather_for_clc aplica el contrato de esta celda:
#   1. Si existe firms_spain_weather.csv (descarga terminada) → se usa tal cual.
#   2. Si no → se parte del parcial y se quedan solo las filas con
#      temperature_2m (lo descargado hasta ahora), para poder avanzar con CLC
#      mientras la descarga sigue en paralelo.
#   3. Si todavía no hay nada de clima → RuntimeError con instrucciones.
#
# [EN] Uses the completed weather CSV if present; otherwise the partial one
#   filtered to the rows that already have temperature_2m.


df_weather = load_weather_for_clc()

print(f"Filas con clima: {len(df_weather):,}")
print(f"Columnas: {df_weather.shape[1]}")
campos_clima = [c for c in weather_fields() if c in df_weather.columns]
print(f"Variables de clima presentes: {len(campos_clima)}/{len(weather_fields())}")
print(f"Filas sin temperature_2m: {df_weather['temperature_2m'].isna().sum()}")

df_weather.head()
```

## 1.4 — Enriquecimiento CLCPlus

Lee el píxel central y los 8 vecinos por detección; escribe `firms_spain_weather_clc.csv`.

```python
# Celda 4 — Enriquecimiento con cobertura del suelo (CLCPlus Backbone) / Land cover
#
# [ESP] Para cada detección lee el píxel central en el tile CLCPlus que le
#   corresponde (EPSG:4326 → EPSG:3035) y, con el barrio activado en
#   configs/project.yaml (clcplus.neighborhood: anillos 1 y 2), también los 8
#   vecinos → columnas clc_class_N/S/W/E/NW/NE/SW/SE y clc_uniform_surroundings.
#   Al cruzar el borde de un tile el vecino se resuelve en el tile contiguo.
#   Se guarda con nombre de etapa firms_spain_weather_clc.csv: cada CSV de
#   data/processed/enriched/ denota su contenido y alimenta al siguiente paso.
#   (data/raw/clcplus/ es solo lectura.)
#
# [EN] Reads the center CLCPlus pixel per detection (plus the 8 ring 1-2
#   neighbors) and saves the result as the named pipeline stage
#   firms_spain_weather_clc.csv, which feeds the next step.


df_clc = enrich_with_clc(df_weather, country="Spain")
ruta_clc = save_enriched(df_clc, filename="firms_spain_weather_clc.csv")

print(f"Filas: {len(df_clc):,}")
print(f"Columnas: {df_clc.shape[1]}")
print("Distribución clc_class:")
print(df_clc["clc_class"].value_counts(dropna=False).sort_index().to_string())
print(f"CSV guardado en: {ruta_clc}")
```

## 1.5 — Filtro decidido: sellado y vegetación

Decisión del equipo (ver celda de código): elimina `clc_class == 1` y conserva `type == 0`; solo filtra en memoria.

```python
# Celda 5 — Filtro decidido: sellado y vegetación / Decided filter
#
# [ESP] Decisión del equipo: se eliminan las filas con clc_class == 1 («Sealed»)
#   y se conservan solo fuegos de vegetación (type == 0). Reduce 47.505 →
#   30.931 filas en firms_spain_full_enrichment.csv.
#
# [EN] Team decision: drop rows with clc_class == 1 («Sealed») and keep
#   vegetation fires only (type == 0). Shrinks 47,505 → 30,931 rows in
#   firms_spain_full_enrichment.csv.

df_clc = df_clc[df_clc["clc_class"] != 1]
df_clc = df_clc[df_clc["type"] == 0]
```

## 1.6 — Región CCAA y presupuestos

Cruza cada detección con el GeoJSON de CCAA y suma presupuestos; escribe `firms_spain_full_enrichment.csv`.

```python
# Celda 6 — Región CCAA + presupuestos / CCAA region + budgets
#
# [ESP] assign_ccaa cruza cada detección con los límites autonómicos
#   (data/raw/ccaa/spain-communities.geojson, fuente decidida por el equipo):
#   1. sjoin "within" en EPSG:3035; los puntos sin polígono que los contenga
#      (costa / frontera PT-FR) usan el respaldo sjoin_nearest con tope
#      ccaa.nearest_max_distance_m = 2000 m; puntos en frontera compartida
#      → se queda con el primer polígono (nunca se duplican filas).
#   2. Los nombres del GeoJSON se traducen al nombre EXACTO del CSV de
#      presupuestos (REGION_NAME_MAP: "Castilla-Leon" → "Castilla y León",
#      "Valencia" → "Comunidad Valenciana"...). Ceuta/Melilla no están
#      en el presupuesto → sus sumas salen NaN.
#   3. add_ccaa_budget_sums → prevention_current (año de la fila) +
#      prevention_prev{N} / extinction_prev{N} con N=ccaa.budget_lookback_years
#      años previos (la extinción excluye el año en curso: sin fuga).
#      (NaN, nunca 0, si la ventana está vacía).
#   4. Guardar firms_spain_full_enrichment.csv → salida completa del
#      enriquecimiento (la entrada del modelado, firms_spain_final.csv,
#      se genera después tras exploración + dedup).
#
# [EN] Joins each detection with the CCAA boundaries: within-join in
#   EPSG:3035, nearest fallback (≤ ccaa.nearest_max_distance_m) for
#   coastal / border points, border ties deduped keep-first, GeoJSON names
#   translated to the exact budget names, then prevention_current (row's
#   year) plus prevention_prev{N} / extinction_prev{N} trailing sums
#   (N=ccaa.budget_lookback_years, current year excluded from extinction:
#   no leakage); saves the full-enrichment
#   output firms_spain_full_enrichment.csv (the modeling input
#   firms_spain_final.csv is produced later, after exploration + dedup).


#
df_full_enrichment = add_ccaa_budget_sums(assign_ccaa(df_clc))
ruta_final = save_enriched(df_full_enrichment, filename="firms_spain_full_enrichment.csv")
#

num_years = load_config()["ccaa"]["budget_lookback_years"]
print(f"Filas: {len(df_full_enrichment):,}")
print(f"Columnas: {df_full_enrichment.shape[1]}")
print("Distribución ccaa:")
print(df_full_enrichment["ccaa"].value_counts(dropna=False).to_string())
print(f"Filas sin ccaa: {int(df_full_enrichment['ccaa'].isna().sum())}")
print(f"Filas sin prevention_current: {int(df_full_enrichment['prevention_current'].isna().sum())}")
print(f"Filas sin prevention_prev{num_years}: {int(df_full_enrichment[f'prevention_prev{num_years}'].isna().sum())}")
print(f"Filas sin extinction_prev{num_years}: {int(df_full_enrichment[f'extinction_prev{num_years}'].isna().sum())}")
print(f"CSV guardado en: {ruta_final}")
#
df_full_enrichment.head()
```

## 2.2 — Estadística descriptiva

Describe el `df_no_dupe` de la 2.1 (sin columnas de contexto); solo visualiza.

```python
# Estadística descriptiva


num_years = load_config()["ccaa"]["budget_lookback_years"]

pd.set_option('display.max_rows', 500)
pd.set_option('display.max_columns', 500)
pd.set_option('display.width', 1000)


DESCRIBE_DROP_COLUMNS = [
    "acq_date",
    "acq_time",
    "latitude",
    "longitude",
    "confidence_cat",
    "confidence_num",
    "daynight",
    "clc_class",
    "clc_class_N",
    "clc_class_S",
    "clc_class_W",
    "clc_class_E",
    "clc_class_NW",
    "clc_class_NE",
    "clc_class_SW",
    "clc_class_SE",
    "clc_uniform_surroundings",
    "ccaa",
    "type",
    "year",
    "is_day",
    "prevention_current",
    f"prevention_prev{num_years}",
    f"extinction_prev{num_years}"
]

df_prov_describe = df_no_dupe.drop(columns=DESCRIBE_DROP_COLUMNS)

print(df_prov_describe.columns)
display(df_prov_describe.describe())
```

## 2.3 — Limpieza de filas sin CCAA ni presupuestos

Elimina filas sin región o sin sumas (Ceuta/Melilla); produce `df_nan_clean`.

```python
# Limpieza de filas sin CCAA y sin presupuestos (Ceuta y Melilla)

display(df_no_dupe.isna().sum())

# Las filas sin ccaa o sin presupuestos (Ceuta/Melilla) se eliminan
# (ver recuentos de faltantes arriba).
df_nan_clean = df_no_dupe.dropna(subset=["ccaa", "prevention_current", f"prevention_prev{num_years}", f"extinction_prev{num_years}"])
```

## 2.4 — Imputación de `boundary_layer_height`

Rellena los faltantes con la mediana; el recuento se verifica en la salida.

```python
# Detección de valores nulos; asignación de valor a boundary_layer_height
# Los faltantes de boundary_layer_height (ver recuento arriba) se imputan
# con la mediana por ser una variable con atípicos.

df_nan_clean.fillna({"boundary_layer_height": df_nan_clean["boundary_layer_height"].median()}, inplace=True)

display(df_nan_clean.isna().sum())

print(df_nan_clean.shape)
```

## 2.5 — Detección de atípicos (IQR)

Regla de Tukey solo como informe; no elimina ni transforma nada.

```python
# Detección de outliers
# Detección valores atipicos / Outlier detection (main dataframe is df_nan_clean)

df_prov_describe = df_nan_clean.drop(columns=DESCRIBE_DROP_COLUMNS)

display(df_prov_describe.describe())

# We apply the Turkey IQR rule
Q1, Q3 = df_prov_describe.quantile(0.25), df_prov_describe.quantile(0.75)
IQR = Q3 - Q1
outlier_counts = ((df_prov_describe < Q1 - 1.5 * IQR) | (df_prov_describe > Q3 + 1.5 * IQR)).sum()
print(outlier_counts[outlier_counts > 0])

# Nuestros outliers son: frp, scan, brightness, brightness_ir, surface_pressure, wind_speed_10m, wind_gusts_10m, boundary_layer_height, vapour_pressure_deficit
# Our outliers are: frp, scan, brightness, brightness_ir, surface_pressure, wind_speed_10m, wind_gusts_10m, boundary_layer_height, vapour_pressure_deficit

# temperature_2m solo tiene 52 valores que se salen de los percentiles, así que no son outliers significativos
# temperature_2m only has 52 values that fall outside the percentiles, so they are not significant outliers.

# precipitation y rain no son outliers ya que la mayoría de los días no llueve (0), por lo que parecen outliers aunque no lo sean
# precipitation and rain are not outliers since most days it does not rain (0), so they seem like outliers even though they are not
```

## 2.6 — Descomposición temporal

`acq_date` a datetime más `month`, `day_of_year` y `es_fin_semana`.

```python
# DESCOMPOSICIÓN TEMPORAL
df_nan_clean['acq_date'] = pd.to_datetime(df_nan_clean['acq_date'])

df_nan_clean['month'] = df_nan_clean['acq_date'].dt.month
df_nan_clean['day_of_year'] = df_nan_clean['acq_date'].dt.dayofyear
df_nan_clean['es_fin_semana'] = df_nan_clean['acq_date'].dt.dayofweek.isin([5, 6]).astype(int)
```

## 2.7 — Tramos horarios de `acq_time`

Hora extraída de `acq_time` y recodificada en `franja_horaria`.

```python
# Segmentación de acq_time en tramos del día
df_nan_clean['acq_time_str'] = df_nan_clean['acq_time'].astype(str).str.zfill(4)
df_nan_clean['hora'] = pd.to_datetime(df_nan_clean['acq_time_str'], format='%H%M', errors='coerce').dt.hour

bins_hora = [-1, 6, 12, 18, 24]
labels_hora = ['Madrugada', 'Mañana', 'Tarde_Pico_Calor', 'Noche']
df_nan_clean['franja_horaria'] = pd.cut(df_nan_clean['hora'], bins=bins_hora, labels=labels_hora, right=False)
```

## 2.8 — Festivos y riesgo antrópico

Cruza fecha×CCAA con el calendario (`holidays.Spain`) en `es_festivo` y `riesgo_antropico_alto`.

```python
# Calcular festivos solo para combinaciones únicas de fecha/CCAA


# Festivos 
mapa_ccaa_holidays = {
    'Andalucía': 'AN', 'Aragón': 'AR', 'Asturias': 'AS', 'Islas Baleares': 'IB', 
    'Canarias': 'CN', 'Cantabria': 'CB', 'Castilla y León': 'CL', 
    'Castilla-La Mancha': 'CM', 'Cataluña': 'CT', 'Comunidad Valenciana': 'VC', 
    'Extremadura': 'EX', 'Galicia': 'GA', 'Madrid': 'MD', 
    'Región de Murcia': 'MC', 'Navarra': 'NC', 'País Vasco': 'PV', 'La Rioja': 'RI'
}

calendario_unico = df_nan_clean[['acq_date', 'ccaa']].dropna().drop_duplicates()

def evaluar_festivo(fila):
    codigo_region = mapa_ccaa_holidays.get(fila['ccaa'])
    if not codigo_region:
        return 0
    cal = holidays.Spain(years=fila['acq_date'].year, subdiv=codigo_region)
    return int(fila['acq_date'] in cal)

calendario_unico['es_festivo'] = calendario_unico.apply(evaluar_festivo, axis=1)

df_nan_clean = df_nan_clean.merge(calendario_unico, on=['acq_date', 'ccaa'], how='left')
df_nan_clean['es_festivo'] = df_nan_clean['es_festivo'].fillna(0).astype(int)

# Variable riesgo antrópico
df_nan_clean['riesgo_antropico_alto'] = ((df_nan_clean['es_fin_semana'] == 1) | (df_nan_clean['es_festivo'] == 1)).astype(int)
```

## 2.9 — Codificación cíclica

Seno/coseno de mes y día del año para la estacionalidad sin saltos.

```python
# Codificación cíclica con sin/cos

df_nan_clean['month_sin'] = np.sin(2 * np.pi * df_nan_clean['month'] / 12)
df_nan_clean['month_cos'] = np.cos(2 * np.pi * df_nan_clean['month'] / 12)
df_nan_clean['day_of_year_sin'] = np.sin(2 * np.pi * df_nan_clean['day_of_year'] / 365.25)
df_nan_clean['day_of_year_cos'] = np.cos(2 * np.pi * df_nan_clean['day_of_year'] / 365.25)
```

## 2.10 — Hectáreas MITECO e índices finales

Ratios por hectárea e índices meteorológicos derivados; escribe `expanded/wildfire_data_cleaned.csv`.

```python
# Hectáreas forestales (MITECO)

num_years = load_config()["ccaa"]["budget_lookback_years"]

superficie_forestal_ha = {
    'Andalucía': 4400000, 'Aragón': 2600000, 'Asturias': 770000,
    'Islas Baleares': 220000, 'Canarias': 570000, 'Cantabria': 360000,
    'Castilla y León': 5100000, 'Castilla-La Mancha': 3600000, 'Cataluña': 2000000,
    'Comunidad Valenciana': 1300000, 'Extremadura': 2700000, 'Galicia': 2000000,
    'Madrid': 430000, 'Región de Murcia': 510000, 'Navarra': 610000,
    'País Vasco': 490000, 'La Rioja': 340000
}

df_nan_clean['sup_forestal_ha'] = df_nan_clean['ccaa'].map(superficie_forestal_ha)

df_nan_clean['ratio_prevencion_extincion'] = df_nan_clean[f'prevention_prev{num_years}'] / (df_nan_clean[f'prevention_prev{num_years}'] + 1e-5)
df_nan_clean['euros_prevencion_por_ha'] = (df_nan_clean[f'prevention_prev{num_years}'] * 1000000) / df_nan_clean['sup_forestal_ha']
df_nan_clean['euros_extincion_por_ha'] = (df_nan_clean[f'extinction_prev{num_years}'] * 1000000) / df_nan_clean['sup_forestal_ha']


# Potencial de secado atmosférico acelerado por el viento
df_nan_clean['potencial_propagacion'] = df_nan_clean['vapour_pressure_deficit'] * df_nan_clean['wind_speed_10m']

df_nan_clean['indice_riesgo_meteo'] = df_nan_clean['potencial_propagacion'] / (df_nan_clean['soil_moisture_0_to_7cm'] + 0.01)

df_nan_clean['estres_termico_vegetacion'] = df_nan_clean['temperature_2m'] / (df_nan_clean['relative_humidity_2m'] + 1)


df_nan_clean['indice_riesgo_meteo_log'] = np.log1p(df_nan_clean['indice_riesgo_meteo'])

# 7. Limpieza final
columnas_basura = [
    'acq_time_str', 'hora', 'sup_forestal_ha', 
    'es_fin_semana', 'es_festivo'
]
df_nan_clean = df_nan_clean.drop(columns=columnas_basura)

save_csv(df_nan_clean, "data/processed/expanded/wildfire_data_cleaned.csv")

display(df_nan_clean)
print(df_nan_clean.shape)
```
